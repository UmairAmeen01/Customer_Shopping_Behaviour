```python
query="""with customer_type as(
  SELECT "Customer ID","Previous Purchases"
  CASE
    WHEN "Previous Purchases" = 1 THEN 'New'
    WHEN "Previous Purchases" BETWEEN 2 AND 10 THEN 'Returning'
    ELSE 'Loyal'
    END AS customer_segment
    from customers

    SELECT customer_segment, count(*) as "Number of customers"
    from customer_type
    GROUP BY customer_segment
    ORDER BY "Number of customers" DESC;
)"""
```

```python
query = """WITH customer_type AS (
  SELECT
    "Customer ID",
    "Previous Purchases",
    CASE
      WHEN "Previous Purchases" = 1 THEN 'New'
      WHEN "Previous Purchases" BETWEEN 2 AND 10 THEN 'Returning'
      ELSE 'Loyal'
    END AS customer_segment
  FROM
    customers
)
SELECT
  customer_segment,
  count(*) AS "Number of customers"
FROM
  customer_type
GROUP BY
  customer_segment
ORDER BY
  "Number of customers" DESC;"""

result = pd.read_sql(query, conn)
print(result)
```

```python
query = """WITH item_counts AS (
    SELECT "Category",
           "Item Purchased",
           COUNT("Customer ID") AS total_orders,
           ROW_NUMBER() OVER (PARTITION BY "Category" ORDER BY COUNT("Customer ID") DESC) AS item_rank
    FROM customers
    GROUP BY category, "Item Purchased"
)
SELECT item_rank,category, "Item Purchased", total_orders
FROM item_counts
WHERE item_rank <=3;"""

result = pd.read_sql(query, conn)
print(result)
```

```python
query="""
SELECT "Subscription Status",
COUNT("Customer ID") AS repeat_buyers
FROM Customers
WHERE  "Previous Purchases" > 5
GROUP BY "Subscription Status";
"""

result = pd.read_sql(query, conn)
print(result)
```

```python
query="""SELECT
    CASE
        WHEN previous_purchases BETWEEN 0 AND 4 THEN '0-4'
        WHEN previous_purchases BETWEEN 5 AND 9 THEN '5-9'
        WHEN previous_purchases BETWEEN 10 AND 14 THEN '10-14'
        WHEN previous_purchases BETWEEN 15 AND 19 THEN '15-19'
        WHEN previous_purchases BETWEEN 20 AND 24 THEN '20-24'
        WHEN previous_purchases BETWEEN 25 AND 29 THEN '25-29'
        WHEN previous_purchases BETWEEN 30 AND 34 THEN '30-34'
        WHEN previous_purchases BETWEEN 35 AND 39 THEN '35-39'
        WHEN previous_purchases BETWEEN 40 AND 44 THEN '40-44'
        WHEN previous_purchases BETWEEN 45 AND 49 THEN '45-49'
        ELSE '50'
    END AS purchase_group,

    AVG(review_rating) AS avg_rating,

    COUNT(DISTINCT customer_id) AS customer_count

FROM customers

GROUP BY purchase_group

ORDER BY
    CASE
        WHEN purchase_group = '0-4' THEN 0
        WHEN purchase_group = '5-9' THEN 5
        WHEN purchase_group = '10-14' THEN 10
        WHEN purchase_group = '15-19' THEN 15
        WHEN purchase_group = '20-24' THEN 20
        WHEN purchase_group = '25-29' THEN 25
        WHEN purchase_group = '30-34' THEN 30
        WHEN purchase_group = '35-39' THEN 35
        WHEN purchase_group = '40-44' THEN 40
        WHEN purchase_group = '45-49' THEN 45
        ELSE 50
    END;"""

result = pd.read_sql(query, conn)
print(result)
```
